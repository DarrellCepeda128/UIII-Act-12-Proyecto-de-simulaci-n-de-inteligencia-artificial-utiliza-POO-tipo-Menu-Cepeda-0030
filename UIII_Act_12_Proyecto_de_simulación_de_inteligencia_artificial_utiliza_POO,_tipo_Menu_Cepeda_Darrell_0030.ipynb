## Ejemplo Carpintero

```python
# --- SISTEMA DE IA: GESTIÓN ORIENTADA A OBJETOS (CARPINTERÍA) ---

class Carpintero:
    # El método __init__ inicializa los atributos del objeto al crearse
    def __init__(self, nombre, madera, maquinaria, catalogo, taller, pedidos, costo):
        self.nombre = nombre
        self.madera = madera
        self.maquinaria = maquinaria  # "si" o "no"
        self.catalogo = catalogo
        self.taller = taller
        self.pedidos = pedidos
        self.costo = costo

    # Función de IA para evaluar capacidad de instalación
    def instalacion(self):
        print(f"\n> IA Analizando Instalación para: {self.nombre}")
        # Lógica de decisión basada en atributos
        if self.maquinaria == "si" and self.costo < 5000:
            print("IA DECISIÓN: Perfil IDEAL para instalaciones rápidas y económicas.")
        elif self.maquinaria == "no":
            print("IA ALERTA: Requiere apoyo de equipo externo para instalar.")
        else:
            print("IA ESTADO: Capacidad estándar para montajes.")

    # Función de IA para evaluar capacidad de reparación
    def reparacion(self):
        print(f"\n> IA Analizando Reparaciones para: {self.nombre}")
        if self.pedidos > 8:
            print("IA DECISIÓN: NO ASIGNAR. Taller saturado de pedidos pendientes.")
        else:
            print("IA DECISIÓN: DISPONIBLE. Capacidad óptima para reparaciones urgentes.")

    # Función de IA para evaluar proyectos de construcción
    def construccion(self):
        print(f"\n> IA Analizando Construcción para: {self.nombre}")
        if self.madera.lower() == "caoba" or self.madera.lower() == "cedro":
            print(f"IA PERFIL: Experto en maderas finas. Asignar construcción de {self.catalogo}.")
        else:
            print(f"IA PERFIL: Especialista en estructuras generales con {self.madera}.")

# Lista para almacenar las instancias (objetos) de la clase Carpintero
base_datos_carpinteros = []

# --- MENÚ INTERACTIVO ---
while True:
    print("\n==============================================")
    print("      CENTRO DE MANDO IA: CARPINTEROS POO     ")
    print("==============================================")
    print("1. Registrar nuevo Carpintero")
    print("2. Ver listado organizado (Tabla)")
    print("3. Ejecutar Diagnóstico de IA (Análisis)")
    print("4. Salir")

    opcion = input("\nSeleccione una opción (1-4): ")

    if opcion == "1":
        print("\n--- ENTRADA DE DATOS DEL OBJETO ---")
        nombre = input("Nombre del carpintero: ")
        madera = input("Tipo de madera preferida: ")
        maquinaria = input("¿Maquinaria propia? (si/no): ").lower()
        catalogo = input("Catálogo principal (muebles): ")
        taller = input("Nombre o ubicación del taller: ")
        pedidos = int(input("Pedidos pendientes actuales: "))
        costo = float(input("Costo de presupuesto promedio ($): "))

        # Instanciación: Se crea el objeto a partir de la clase
        nuevo_carpintero = Carpintero(nombre, madera, maquinaria, catalogo, taller, pedidos, costo)
        base_datos_carpinteros.append(nuevo_carpintero)
        print(f"\nIA: Objeto '{nombre}' creado y guardado en memoria.")

    elif opcion == "2":
        if not base_datos_carpinteros:
            print("\nIA ALERTA: No hay objetos registrados en la lista.")
        else:
            print("\n" + "-" * 115)
            # Encabezado de tabla alineado con f-strings
            header = f"{'Nombre':<15} | {'Madera':<12} | {'Máq.':<6} | {'Pedidos':<8} | {'Presupuesto':<12} | {'Taller':<15}"
            print(header)
            print("-" * 115)

            # Recorrido de la lista de objetos para mostrar sus atributos
            for carp in base_datos_carpinteros:
                print(f"{carp.nombre:<15} | {carp.madera:<12} | {carp.maquinaria:^6} | {carp.pedidos:^8} | ${carp.costo:<11.2f} | {carp.taller:<15}")

    elif opcion == "3":
        if not base_datos_carpinteros:
            print("\nIA ERROR: No hay datos para procesar el diagnóstico.")
        else:
            print("\n--- PROCESANDO COMPORTAMIENTO DE IA ---")
            for carp in base_datos_carpinteros:
                # Se envían mensajes (llamadas a funciones) a los objetos
                carp.instalacion()
                carp.reparacion()
                carp.construccion()

    elif opcion == "4":
        print("\nCerrando sistema de IA... ¡Seguridad y precisión en la obra!")
        break # Finaliza el ciclo while True

    else:
        print("Opción no válida. Intente de nuevo.")

print("------------------------------------------")
print("Simulación de proceso lógico finalizada.")

print("Programa realizado por Cepeda Darrell NL = 10")
```
