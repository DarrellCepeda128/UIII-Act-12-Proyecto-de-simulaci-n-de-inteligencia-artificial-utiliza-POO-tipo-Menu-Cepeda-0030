## Ejemplo Carpintero

```python
# --- SISTEMA DE IA: GESTIÓN DE CARPINTERÍA (POO ALINEADO) ---

class Carpintero:
    # Atributo de clase para almacenar los objetos creados
    base_datos = []

    def __init__(self, nombre, madera, maquinaria, catalogo, taller, pedidos, costo):
        # Inicialización de atributos del objeto
        self.nombre = nombre
        self.madera = madera
        self.maquinaria = maquinaria  # "si" o "no"
        self.catalogo = catalogo
        self.taller = taller
        self.pedidos = pedidos
        self.costo = costo

    def obtener_diagnostico_ia(self):
        # Lógica de IA: evalúa el estado del objeto para decidir
        if self.pedidos > 8: return "IA: SATURADO. Taller con exceso de trabajo."
        elif self.maquinaria == "no" and self.pedidos > 3: return "IA: RIESGO. Falta equipo para cumplir."
        elif self.costo < 3000: return "IA: ECONÓMICO. Ideal para presupuestos bajos."
        else: return "IA: ÓPTIMO. Listo para nuevos proyectos."


    def agregar_carpintero():
        print("\n--- REGISTRO DE NUEVO CARPINTERO ---")
        nom = input("Nombre: "); mad = input("Madera preferida: ")
        maq = input("¿Maquinaria propia? (si/no): ").lower()
        cat = input("Catálogo (Muebles): "); tal = input("Taller (Ubicación): ")
        ped = int(input("Pedidos pendientes: ")); cos = float(input("Costo promedio: "))
        # Se crea la instancia y se agrega a la lista alineada
        nuevo = Carpintero(nom, mad, maq, cat, tal, ped, cos)
        Carpintero.base_datos.append(nuevo)
        print(f"SISTEMA: Objeto '{nom}' creado en memoria.")


    def listar_carpinteros():
        if not Carpintero.base_datos: print("\nIA ALERTA: No hay datos registrados.")
        else:
            print("\n" + "=" * 120)
            print(f"{'Nombre':<15} | {'Madera':<12} | {'Máq.':<5} | {'Taller':<15} | {'Pedidos':<8} | {'Costo':<10} | {'Diagnóstico IA'}")
            print("-" * 120)
            for c in Carpintero.base_datos:
                diag = c.obtener_diagnostico_ia()
                print(f"{c.nombre:<15} | {c.madera:<12} | {c.maquinaria:^5} | {c.taller:<15} | {c.pedidos:^8} | ${c.costo:<9.2f} | {diag}")
            print("=" * 120)


    def actualizar_carpintero():
        nombre_buscar = input("\nCarpintero a modificar: ")
        for c in Carpintero.base_datos:
            if c.nombre.lower() == nombre_buscar.lower():
                c.pedidos = int(input("Nuevos pedidos: "))
                c.costo = float(input("Nuevo costo: "))
                print("SISTEMA: Datos actualizados."); return
        print("SISTEMA: No encontrado.")


    def borrar_carpintero():
        nombre_buscar = input("\nNombre del carpintero a eliminar: ")
        for i in range(len(Carpintero.base_datos)):
            if Carpintero.base_datos[i].nombre.lower() == nombre_buscar.lower():
                Carpintero.base_datos.pop(i)
                print(f"SISTEMA: Objeto '{nombre_buscar}' eliminado."); return
        print("SISTEMA: No encontrado.")

# --- MENÚ INTERACTIVO PRINCIPAL ---
while True:
    print("\n==============================================")
    print("   CENTRO DE MANDO IA: CARPINTEROS (POO)   ")
    print("==============================================")
    print("1. Agregar | 2. Listar | 3. Actualizar | 4. Borrar | 5. Salir")
    opcion = input("\nSeleccione una opción (1-5): ")

    if opcion == "1": Carpintero.agregar_carpintero()
    elif opcion == "2": Carpintero.listar_carpinteros()
    elif opcion == "3": Carpintero.actualizar_carpintero()
    elif opcion == "4": Carpintero.borrar_carpintero()
    elif opcion == "5": print("\nCerrando sesión de IA..."); break
    else: print("Error: Opción no válida.")

print("Programa realizado por Cepeda Darrell NL = 10")
```
