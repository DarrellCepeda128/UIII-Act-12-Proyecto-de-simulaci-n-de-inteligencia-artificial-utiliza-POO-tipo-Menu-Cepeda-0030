## Ejemplo Carpintero

```python
# --- SISTEMA DE IA: GESTIÓN DE CARPINTERÍA (POO) ---

class Carpintero:
    """Clase que representa a un carpintero y sus capacidades"""

    def __init__(self, nombre, madera, maquinaria, catalogo, taller, pedidos, costo):
        # Inicialización de atributos del objeto
        self.nombre = nombre
        self.madera = madera
        self.maquinaria = maquinaria  # "si" o "no"
        self.catalogo = catalogo
        self.taller = taller
        self.pedidos = pedidos
        self.costo = costo

    def obtener_diagnostico_ia(self):
        """Simula una decisión inteligente basada en los atributos del objeto"""
        if self.pedidos > 8:
            return "IA: SATURADO. No asignar más pedidos por ahora."
        elif self.maquinaria == "no" and self.pedidos > 3:
            return "IA: RIESGO. Requiere maquinaria para cumplir plazos."
        elif self.costo < 3000 and self.madera.lower() == "pino":
            return "IA: COMPETITIVO. Ideal para proyectos de bajo costo."
        else:
            return "IA: ÓPTIMO. Taller listo para recibir proyectos nuevos."

# --- FUNCIONES DE GESTIÓN DEL SISTEMA ---
# Se utiliza una lista para almacenar los objetos (instancias)
base_datos_carpinteros = []

def agregar_carpintero():
    print("\n--- REGISTRO DE NUEVO CARPINTERO ---")
    nombre = input("Nombre del carpintero: ")
    madera = input("Tipo de madera preferida: ")
    maquinaria = input("¿Maquinaria propia? (si/no): ").lower()
    catalogo = input("Catálogo de muebles (ej. Sillas, Mesas): ")
    taller = input("Nombre o ubicación del taller: ")
    pedidos = int(input("Número de pedidos pendientes: "))
    costo = float(input("Costo de presupuesto promedio ($): "))

    # Instanciación: Creación del objeto a partir de la clase
    nuevo = Carpintero(nombre, madera, maquinaria, catalogo, taller, pedidos, costo)
    base_datos_carpinteros.append(nuevo)
    print(f"\nSISTEMA: Objeto '{nombre}' creado y guardado en memoria.")

def listar_carpinteros():
    if not base_datos_carpinteros:
        print("\nIA ALERTA: La base de datos está vacía.")
    else:
        print("\n" + "=" * 125)
        # Encabezado de tabla bien alineado usando f-strings
        header = f"{'Nombre':<15} | {'Madera':<12} | {'Máq.':<5} | {'Taller':<15} | {'Pedidos':<8} | {'Presupuesto':<12} | {'Diagnóstico IA'}"
        print(header)
        print("-" * 125)

        # Ciclo for para recorrer la lista de objetos
        for c in base_datos_carpinteros:
            # Acceso a los atributos mediante notación de punto
            diagnostico = c.obtener_diagnostico_ia()
            print(f"{c.nombre:<15} | {c.madera:<12} | {c.maquinaria:^5} | {c.taller:<15} | {c.pedidos:^8} | ${c.costo:<11.2f} | {diagnostico}")
        print("=" * 125)

def actualizar_carpintero():
    if not base_datos_carpinteros:
        print("\nSISTEMA: No hay datos para actualizar.")
        return

    nombre_buscar = input("\nIngrese el nombre del carpintero a actualizar: ")
    encontrado = False

    for c in base_datos_carpinteros:
        if c.nombre.lower() == nombre_buscar.lower():
            print(f"Modificando datos de {c.nombre}...")
            c.pedidos = int(input("Ingrese nueva cantidad de pedidos: "))
            c.costo = float(input("Ingrese nuevo costo de presupuesto: "))
            print("SISTEMA: Datos actualizados correctamente.")
            encontrado = True
            break # Sale del ciclo una vez encontrado

    if not encontrado:
        print("SISTEMA: No se encontró al carpintero.")

def borrar_carpintero():
    if not base_datos_carpinteros:
        print("\nSISTEMA: No hay datos para borrar.")
        return

    nombre_buscar = input("\nNombre del carpintero a eliminar: ")
    for i in range(len(base_datos_carpinteros)):
        # Accedemos al objeto en la posición i para comparar el nombre
        if base_datos_carpinteros[i].nombre.lower() == nombre_buscar.lower():
            # Eliminamos el objeto de la lista
            base_datos_carpinteros.pop(i)
            print(f"SISTEMA: Objeto '{nombre_buscar}' eliminado de la memoria.")
            return
    print("SISTEMA: Carpintero no encontrado.")

# --- MENÚ INTERACTIVO (CICLO PRINCIPAL) ---
while True:
    print("\n==============================================")
    print("   CENTRO DE MANDO IA: CARPINTEROS (POO)   ")
    print("==============================================")
    print("1. Agregar Carpintero")
    print("2. Listar Carpinteros (Ver Tabla e IA)")
    print("3. Actualizar Carpintero")
    print("4. Borrar Carpintero")
    print("5. Salir")

    opcion = input("\nSeleccione una opción (1-5): ")

    if opcion == "1":
        agregar_carpintero()
    elif opcion == "2":
        listar_carpinteros()
    elif opcion == "3":
        actualizar_carpintero()
    elif opcion == "4":
        borrar_carpintero()
    elif opcion == "5":
        print("\nCerrando sesión de IA. ¡Seguridad en el taller!")
        break # Finaliza el ciclo while True
    else:
        print("Error: Opción no válida.")

print("Simulación de proceso lógico finalizada.")

print("Programa realizado por Cepeda Darrell NL = 10")
```
